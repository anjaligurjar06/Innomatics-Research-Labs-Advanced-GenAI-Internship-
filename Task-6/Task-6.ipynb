# Problem Statement 1:Smart Parking Lot Management System

```python
def parking_system(limit, logs):
    
    total = 0
    peak = 0

    for i in logs:
        if i == "IN":
            total += 1
        elif i == "OUT":
            total -= 1

        if total > peak:
            peak = total

    print("Currently Parked Vehicles:", total)
    print("Peak Parking Usage:", peak)

    if total > limit:
        print("Parking Status: Capacity Exceeded")
    else:
        print("Parking Status: Available")
limit = int(input("Enter Parking Capacity: "))
# Use eval to convert string list to real list
logs = eval(input("Enter Vehicle Logs (IN/OUT): "))

parking_system(limit, logs)
```

# Problem Statement 2:Online Food Delivery Time Estimator

```python
def estimate_delivery(distance, traffic, weather):
    speed = 20
    
    # Calculate base time in minutes
    time = (distance / speed) * 60
    
    # Traffic delay
    if traffic == "Low":
        time += 5
    elif traffic == "Medium":
        time += 10
    elif traffic == "High":
        time += 20
    
    # Weather delay
    if weather == "Rainy":
        time += 10
    elif weather == "Stormy":
        time += 20

    print("Estimated Delivery Time:", int(time), "minutes")

distance = int(input("Distance (km): "))
traffic = input("Traffic Level: ")
weather = input("Weather: ")

estimate_delivery(distance, traffic, weather)
```

# Problem Statement 3:Movie Theatre Seat Occupancy 

```python
# Function to analyze seat booking
def analyze_seats(total_seats, booked_list):
     # Count how many seats are booked
    booked_count = len(booked_list)
    
    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100
    
    print("Total Booked Seats:", booked_count)
    print("Occupancy Percentage:", round(occupancy, 2), "%")
    if booked_count == total_seats:
        print("Status: Housefull")
    else:
        print("Status: Seats Available")
    if occupancy >= 80:
        print("Suggestion: Consider Opening Additional Show")
    else:
        print("Suggestion: No Additional Show Needed")

total_seats = int(input("Total Seats: "))

# For testing purpose (simulate 150 bookings)
booked_list = [1] * 150  

analyze_seats(total_seats, booked_list)
```

# Problem Statement 4:Cloud Server Load Classification System

```python
# Function to classify server load
def classify_server(cpu_readings):
    total = 0  
    for i in cpu_readings:
        total += i
    average = total / len(cpu_readings)
    
    print("Average CPU Load:", int(average), "%")
    
    # Classify based on average
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

cpu_readings = [45, 60, 70, 85, 90]

classify_server(cpu_readings)
```

# Problem Statement 5:Smart Classroom Resource Usage Monitor

```python
def check_resource_usage(usage_dict):
    
    overused = [] 
    for resource in usage_dict:
        hours = usage_dict[resource]
        # Check if usage exceeds 8 hours
        if hours > 8:
            overused.append(resource)
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

check_resource_usage(usage_data)
```

#  Problem Statement 6: Online Event Registration Capacity Controller

```python
# Function to manage event registrations
def manage_registration(capacity, total_requests):
    
    confirmed = 0
    waitlist = 0
    for i in range(total_requests):
        # If seats available
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlist += 1
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    if confirmed == capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")

capacity = int(input("Event Capacity: "))
registrations = int(input("Registrations: "))

manage_registration(capacity, registrations)
```
