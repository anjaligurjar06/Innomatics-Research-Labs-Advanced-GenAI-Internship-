# Problem 1: Social Media – Post Engagement Analyzer

```python
def analyse_engagement(likes):
    total_likes = 0  
    # loop to calculate total likes
    for i in likes:
        total_likes += i   # adding each like to total_likes
    print("Total Likes:", total_likes)
    # checking if post is viral or normal
    if total_likes >= 1000:
        print("Post Status: Viral Post")
    else:
        print("Post Status: Normal Engagement")

likes = [100, 100, 200, 300, 400, 150]
# calling the function
analyse_engagement(likes)
```

# Problem 3: Agriculture – Rainfall Adequacy Checker


```python
def analyse_rainfall(rainfall,required):
    total_rainfall=0
    for i in rainfall:
        total_rainfall+=i
    avg_rainfall=total_rainfall/len(rainfall)
    print("Average Rainfall:", avg_rainfall)
    if avg_rainfall >= required:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")
rainfall=[100,45,78,99,34,78,58,64]
required=59
analyse_rainfall(rainfall,required)

    
```

# Problem 4: Social Media – Duplicate Account Detection

```python
# Function to detect duplicate usernames
def detect_duplicates(usernames):
    # converting list to set (set removes duplicates automatically)
    unique_usernames = set(usernames)
    # checking if duplicates exist
    if len(usernames) != len(unique_usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

usernames = ["anjali", "akshu", "diksha", "diksha", "anish","sarthak","sahil"]
detect_duplicates(usernames)
```

# Problem 6: Agriculture – Premium Crop Price Filter

```python
def crop_filter(prices):
    premium_crops = []   # list to store premium crop prices
    for price in prices:
        if price > 2000:      
            premium_crops.append(price)
    print("Premium Crops:", premium_crops)
prices = [1500, 2500, 1800, 3200, 2000]

# calling the function
crop_filter(prices)
```

# Problem 7: System Monitoring – Application Health

```python
def application_health_checker(errors):
    print("Error Count:", errors)
    
    # checking system status based on error count
    if errors == 0:
        print("System Status: Healthy")
    elif errors <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")
errors = 7

# calling the function
application_health_checker(errors)
```

# Problem 8: Banking – Daily Transaction Limit Checker

# Problem 9: E-Learning – Student Attendance Eligibility System


```python
def attendance_checker(attendance_list):
    total_classes = len(attendance_list)   
    attended = 0
    for i in attendance_list:
        attended += i
    attendance_percentage = (attended / total_classes) * 100
    print("Attendance Percentage:", attendance_percentage)
    if attendance_percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")

attendance_list = [1,1,1,1,0,1,1,1,0,1] 

# calling the function
attendance_checker(attendance_list)
```

# Problem 10: Smart Electricity Bill Analyzer

```python
def electricity_bill_analyzer(units):
    bill = 0 
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + (units - 100) * 5
    else:
        bill = (100 * 3) + (100 * 5) + (units - 200) * 7
    if bill < 500:
        usage_status = "Low Usage"
    elif bill <= 1500:
        usage_status = "Moderate Usage"
    else:
        usage_status = "High Usage"
    
    return bill, usage_status

units = 250

total_bill, status = electricity_bill_analyzer(units)

print("Units Consumed:", units)
print("Total Bill:", total_bill)
print("Usage Level:", status)
```
